```python
import pandas as pd
import sqlite3

matches = pd.read_csv('matches.csv')

conn = sqlite3.connect(':memory:')
matches.to_sql('matches', conn, index=False)

query1 = """
SELECT match_winner, COUNT(*) as wins
FROM matches
GROUP BY match_winner
ORDER BY wins DESC
"""

result = pd.read_sql(query1, conn)
print(result)
```

# Toss Analysis

```python
query2 = """
SELECT toss_decision, COUNT(*) as times
FROM matches
GROUP BY toss_decision
"""
result2 = pd.read_sql(query2, conn)
print(result2)
```

# Top venue

```python
query3 = """
SELECT venue, COUNT(*) as matches_played
FROM matches
GROUP BY venue
ORDER BY matches_played DESC
LIMIT 5
"""
result3 = pd.read_sql(query3, conn)
print(result3)
```

# Toss Winner=Match Winner

```python
query4 = """
SELECT 
    SUM(CASE WHEN toss_winner = match_winner THEN 1 ELSE 0 END) as toss_jeeto_match_jeeto,
    SUM(CASE WHEN toss_winner != match_winner THEN 1 ELSE 0 END) as toss_jeeto_match_nahi
FROM matches
"""
result4 = pd.read_sql(query4, conn)
print(result4)
```
