<H1> IMPORT FILES 

```python
## import pandas as pd
import matplotlib.pyplot as plt

# This files can loaded it.
matches = pd.read_csv('matches.csv')
batting = pd.read_csv('batting_stats.csv')
bowling = pd.read_csv('bowling_stats.csv')
points = pd.read_csv('points_table.csv')

# Access each file shape
print("Matches:", matches.shape)
print("Batting:", batting.shape)
print("Bowling:", bowling.shape)
print("Points:", points.shape)

# Display Columns
print("\nMatches columns:",matches.columns.tolist())
print("Batting columns:",batting.columns.tolist())
print("Bowling columns:",bowling.columns.tolist())
print("Points columns:",points.columns.tolist())
```

<h1> DATA CLEANING

```python
# Cleaning All Files

matches.drop_duplicates(inplace=True)
batting.drop_duplicates(inplace=True)
bowling.drop_duplicates(inplace=True)
points.drop_duplicates(inplace=True)

print("Nulls in Matches:\n",matches.isnull().sum())
print("\nNulls in Batting:\n",batting.isnull().sum())
print("\nNulls in Bowling:\n",bowling.isnull().sum())

print("\n✅Cleaning done!")
```

<h1>TOP 10 BATSMAN

```python
import matplotlib.pyplot as plt

# TOP 10 batsman by runs
top_bat = batting.groupby('batsman')['runs'].sum()
top_bat = top_bat.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_bat.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('IPL 2026 - Top 10 Batsman By Runs', fontsize=14)
plt.xlabel('Player')
plt.ylabel('Total Runs')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('top_batsmen.png', bbox_inches='tight', dpi=150)
plt.show()
```

<h1> TOP 10 BOWLERS</h1>

```python
# Top 10 Bowlers By Wickets
top_bowl = bowling.groupby('bowler')['wickets'].sum()
top_bowl = top_bowl.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_bowl.plot(kind='bar', color='coral', edgecolor='black')
plt.title('IPL 2026 - Top 10 Bowlers By Wickets', fontsize=14)
plt.xlabel('Player')
plt.ylabel('Total Wickets')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('top_bowlers.png', bbox_inches='tight', dpi=150)
plt.show()
```

<h1>TEAM WINS </h1>

```python
# Team Wise Wins
team_wins = matches['match_winner'].value_counts()

plt.figure(figsize=(10,5))
team_wins.plot(kind='bar', color='green', edgecolor='black')
plt.title('IPL 2026 - TEAM WINS', fontsize=14)
plt.xlabel('Team')
plt.ylabel('Wins')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('team_wins.png', bbox_inches='tight', dpi=150)
plt.show()
                          
```

<H1>TOSS ANALYSIS

```python
# TOSS WIN = MATCH WIN
matches['toss_win'] = (
    matches['toss_winner'] == matches['match_winner']
)
toss = matches['toss_win'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(toss,
        labels=['Toss Win = Match win', 'Toss Win != Match Win'],
        autopct='%1.1f%%',
        colors=['#2ecc71','#e74c3c'])
plt.title('Toss vs Match Result - IPL 2026')
plt.savefig('Toss.png', bbox_inches='tight', dpi=150)
plt.show()
```

<H1> POINTS TABLE CHART

```python
# Points table visualization
pts = points.sort_values('points', ascending=True)

plt.figure(figsize=(8,6))
plt.barh(pts['team'], pts['points'], color='purple', edgecolor='black')
plt.title('IPL 2026 - Final Points Table', fontsize=14)
plt.xlabel('Points')
plt.tight_layout()
plt.savefig('points_table.png', bbox_inches='tight', dpi=150)
plt.show()
```
